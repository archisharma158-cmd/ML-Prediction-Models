# Spam Email Detection Model

🔹 Step 1: Dataset Preparation
Collected and loaded the email dataset and examined the available messages and their labels.

🔹 Step 2: Data Preprocessing
Cleaned the text data by handling unnecessary characters, converting text into a suitable format, and preparing it for model training.

🔹 Step 3: Feature Extraction
Converted the email text into numerical features using TF-IDF (Term Frequency–Inverse Document Frequency) so that the Machine Learning model could process the textual data.

🔹 Step 4: Train-Test Split
Divided the dataset into training and testing sets to train the model and evaluate its performance on unseen emails.

🔹 Step 5: Model Training
Trained a Machine Learning classification model on the processed email data to identify patterns associated with spam messages.

🔹 Step 6: Model Evaluation
Evaluated the model using classification metrics such as accuracy, precision, recall, and F1-score to understand how effectively it distinguishes spam from legitimate emails.

🔹 Step 7: Prediction
Tested the trained model with new email messages and generated predictions indicating whether an email is Spam or Ham.

🔹 Step 8: Final Result
The project demonstrates how Natural Language Processing (NLP) and Machine Learning can be combined to build a practical spam detection system.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
raw_data=pd.read_csv(r"C:\Users\DELL\Desktop\email_spam detection\mail_data (2).csv")
```

```python
raw_data['Label'] = raw_data['Category'].map({'ham': 0, 'spam': 1})
```

```python
X = raw_data['Message']
Y = raw_data['Label']
```

```python
print(X)
```

```python
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=3)
```

```python
print(X.shape)
print(X_train.shape)
print(X_test.shape)
```

Model Trained with Logistic Regression

```python
model = LogisticRegression()
vectorizer = TfidfVectorizer()
X_train_features = vectorizer.fit_transform(X_train)
X_test_features = vectorizer.transform(X_test)
model.fit(X_train_features, Y_train)

predictions = model.predict(X_test_features)
print("Accuracy:", accuracy_score(Y_test, predictions))
```

```python
train_predictions = model.predict(X_train_features)
print("Training Accuracy:", accuracy_score(Y_train, train_predictions))
print(train_predictions)
```

```python
predictions = model.predict(X_test_features)

test_results = pd.DataFrame({
    'Message': X_test,
    'Actual': Y_test,
    'Predicted': predictions
})

print("Test predictions:")
print(test_results.head(20))
print("\nTest accuracy:", accuracy_score(Y_test, predictions))
```

```python
def predict_message(message):
    message_features = vectorizer.transform([message])
    prediction = model.predict(message_features)[0]
    return "spam" if prediction == 1 else "ham"

# Example usage
sample_messages = [
    "Congratulations! You've won a $1000 gift card. Click here to claim now.",
    "Hey, can we reschedule our meeting to tomorrow?"
]

for msg in sample_messages:
    print(f"Message: {msg}")
    print(f"Prediction: {predict_message(msg)}\n")
```
